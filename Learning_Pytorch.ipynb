```python
#Libraries
import pandas as pd
import numpy as np
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler , LabelEncoder
from sklearn.metrics import accuracy_score
```

```python
#read csv
df = pd.read_csv('https://raw.githubusercontent.com/gscdit/Breast-Cancer-Detection/refs/heads/master/data.csv')
df.head()
```

```python
df.drop(columns=['id' , 'Unnamed: 32'] , inplace=True)
```

```python
set(df['diagnosis'])
```

```python
all_features = df.iloc[: , 1:]      #all rows , 1st index column to last index (skipping 0th index)
target_variables = df.iloc[: , 0]  #all rows , 0th index column only
X_train , X_test , Y_train , Y_test = train_test_split(all_features , target_variables , test_size=0.3)
```

```python
#standardizing input features
scalar = StandardScaler()
X_train = scalar.fit_transform(X_train)
X_test_scaled = scalar.transform(X_test)
```

```python
X_train.shape
X_test_scaled.shape
```

```python
Y_test
```

```python
#changing the categorical variables into numeric variables
encoder = LabelEncoder()
Y_train = encoder.fit_transform(Y_train)
Y_test_encoded = encoder.transform(Y_test)
```

```python
Y_train
```

```python
# Converting arrays to tensors
X_train_tensor = torch.from_numpy(X_train)
X_test_tensor = torch.from_numpy(X_test_scaled)
Y_train_tensor = torch.from_numpy(Y_train)
Y_test_tensor = torch.from_numpy(Y_test_encoded)
```

```python
X_train_tensor.shape
```

```python
#Neural Network (1 neuron)
class neuralnetwork():
  def __init__(self,X):
    '''
    Weights and Biases
    Since each feature has a single weight, we only need x.size number of rows & just a single column
    Since we are creating a network with a single neuron, we will only have a single bias only aswell
    '''
    self.weights = torch.rand(size=(X.size()[1], 1) ,dtype=X.dtype, requires_grad=True)#initializing random weights (1 weight for each feature)
    self.biases = torch.zeros(size=(1,),dtype=X.dtype, requires_grad=True)   #bias will start from 0

  def forward_pass(self, X):
    '''
    We need to perform a matrix multiplication between the incoming features and the weights and then add the bias terms
    '''
    #Multiplication and Addition
    Z = torch.matmul(X,self.weights) + self.biases   # (featuresMatrix * WeightsMatrix) + biasesMatrix
    #Activation Function
    ypred = torch.sigmoid(Z)
    return ypred    #the size of this tensor would be the same as the size of the original Y column (thus helping in calculation of loss)

  def calculate_loss(self,y_predicted,y_original):
    '''
    Calculating the loss value in order to check how far predicted output is from the original
    Predicted output would be the ypred calculated in the forward pass
    e.g Binary cross entropy loss
    '''
    epsilon = 1e-7
    y_predicted = torch.clamp(y_predicted, epsilon, 1 - epsilon)
    return torch.nn.functional.binary_cross_entropy(y_predicted , y_original)

```

```python
model = neuralnetwork(X_train_tensor)
```

```python
#variables
epochs = 30
learningRate = 0.1
resized_Y_train_tensor = Y_train_tensor.view([-1,1]).double()   #resizing the tensor to look like the prediction tensor and chaing its dtype to be same as the preicted tensor aswell
#loop for training the network
for counter in range(epochs):
  #forward pass
  y_pred = model.forward_pass(X_train_tensor)
  #calculate loss
  loss = model.calculate_loss(y_pred , resized_Y_train_tensor)
  print(f"Epoch: {counter+1} , Loss: {loss.item()}")
  #backward prop
  loss.backward()    #autograd calculates the gradient descent
  #update the weights and biases manually
  with torch.no_grad():   #have to turn off gradient tracking when auto updating weights and biases
    model.weights.data -= learningRate*model.weights.grad
    model.biases.data -= learningRate*model.biases.grad
  #we want the gradients to be calculated from fresh each time in each iteration
  model.weights.grad.zero_()
  model.biases.grad.zero_()
```

```python
model.biases
```

```python
# model evaluation
with torch.no_grad():
  y_pred = model.forward_pass(X_test_tensor)
  y_pred = (y_pred > 0.9).float()
  accuracy = (y_pred == Y_test_tensor).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```

# Creating the same network using nn.Module

```python
from torch import nn
from torch.optim import SGD
class upgradedNeuralNetwork(nn.Module):
  def __init__(self, input_features):
    #initializing the parent constructor
    super().__init__()
    #Creating the input layer having single neuron as first
    self.firstlayer = nn.Linear(input_features , 1)     #the input features are the inputs going into each neuron, the second argument is the number of neurons in this layer
                                                        #the number of outputs from this layer is equivalent to number of neurons
  def forward(self , X):
    #performing the matmul step thorugh the nn.Linear function
    Z = self.firstlayer(X)
    y_pred = torch.sigmoid(Z)
    return y_pred

  def loss(self,y_prediction , y_original):
    return torch.nn.BCELoss()(y_prediction , y_original)
```

```python
#variables
input_features = X_train_tensor.size()[1]

model = upgradedNeuralNetwork(input_features)
epochs = 30
learningrate = 0.01
optimizer = SGD(params=model.parameters() , lr=learningrate)

for counter in range(epochs):
  optimizer.zero_grad()   #reset the gradients
  y_pred = model(X_train_tensor.to(dtype=torch.float32))    #forward pass
  loss = model.loss(y_pred , Y_train_tensor.view([-1,1]).to(dtype=torch.float32))   #calculate the loss value
  print(f"Epochs: {counter+1} , Loss: {loss.item()}")

  loss.backward()     #calculate the new gradients
  optimizer.step()    #updates the weights
```

```python
# model evaluation
with torch.no_grad():
  y_pred = model.forward(X_test_tensor.to(dtype=torch.float32))
  y_pred = (y_pred > 0.9).float()
  accuracy = (y_pred == Y_test_tensor).float().mean()
  print(f'Accuracy: {accuracy.item()}')
```
